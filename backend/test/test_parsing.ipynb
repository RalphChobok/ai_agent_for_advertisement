#### Chunking

```python
from unstructured.partition.pdf import partition_pdf
from unstructured.chunking.title import chunk_by_title

# 1. Extract visual elements
elements = partition_pdf(filename="reference_pack.pdf")

# 2. Chunk by title (sections) automatically respects visual hierarchy
chunks = chunk_by_title(elements)

for i in range(len(chunks)):
    print(f"Chunk{i}: {chunks[i].text[:100]}...\n")
```

```python
from unstructured.partition.pdf import partition_pdf
from unstructured.chunking.title import chunk_by_title
from unstructured.cleaners.core import clean, clean_extra_whitespace

# 1. Extract visual elements
elements = partition_pdf(filename="reference_pack.pdf")

# Optional but recommended: Clean up weird PDF line breaks and whitespace before chunking
for element in elements:
    if hasattr(element, 'text'):
        element.text = clean_extra_whitespace(clean(element.text, dashes=True, bullets=True))

# 2. Enhanced chunking with overlap and boundaries
chunks = chunk_by_title(
    elements,
    combine_text_under_n_chars=500,  # Groups tiny fragments (like single list items) together
    max_characters=1500,             # Expands the upper limit to keep Avatar profiles intact
    new_after_n_chars=1200,          # Soft limit: starts looking for a clean break after this point
    overlap=200,                     # Creates a ~15% overlap to carry context forward
    overlap_all=True                 # Applies the overlap rule across all chunk transitions
)

# 3. Output validation
for i, chunk in enumerate(chunks):
    # Printing the length alongside the text helps you monitor the distribution
    print(f"Chunk{i} [Len: {len(chunk.text)}]: {chunk.text[:100]}...\n")
```
